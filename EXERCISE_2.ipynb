```python
year = int(input("Enter a year: "))

if year % 400 == 0:
    print("Leap year")
elif year % 100 == 0:
    print("Not a leap year")
elif year % 4 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```

```python
number = int(input("Enter a number: "))
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

print("Reverse number:", reverse)
```

```python
ch = input("Enter a character: ")

if ch == 'a' or ch == 'e' or ch == 'i' or ch == 'o' or ch == 'u':
    print("Vowel")
elif ch == 'A' or ch == 'E' or ch == 'I' or ch == 'O' or ch == 'U':
    print("Vowel")
else:
    print("Consonant")
```

```python
name = input("Enter customer name: ")
meter = input("Enter meter number: ")
old = int(input("Enter previous reading: "))
new = int(input("Enter current reading: "))

units = new - old

if units <= 50:
    bill = units * 2.30
elif units <= 200:
    bill = (50 * 2.30) + (units - 50) * 4.00
elif units <= 400:
    bill = (50 * 2.30) + (150 * 4.00) + (units - 200) * 5.00
else:
    bill = (50 * 2.30) + (150 * 4.00) + (200 * 5.00) + (units - 400) * 5.40

print("Customer name:", name)
print("Meter number:", meter)
print("Units used:", units)
print("Bill amount: Rs.", bill)
```

```python
number = int(input("Enter a number: "))
factorial = 1

for i in range(1, number + 1):
    factorial = factorial * i

print("Factorial:", factorial)
```
